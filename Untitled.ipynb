```python
import os
import numpy as np
from tqdm import tqdm
```

```python
NEW_EMBED_DIR = 'new_embeddings/Chinese'
OUTPUT_DIR = 'new_embeddings/'
```

```python
data = []

for file in tqdm(os.listdir(NEW_EMBED_DIR)):
    if not file.endswith('.npy'):
        continue
    else :
        embd = np.load(f"{NEW_EMBED_DIR}{file}")
        data.append(embd)
```
