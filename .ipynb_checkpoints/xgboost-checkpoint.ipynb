```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import xgboost as xgb
import os
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from xgboost import XGBClassifier

%matplotlib inline
sns.set_style("whitegrid")
```

```python
#so taking all the embedding and loading them

df_ch1 = pd.read_csv('embeddings/chinese1.csv')
df_ch2 = pd.read_csv('embeddings/chinese2.csv')

df_ko1 = pd.read_csv('embeddings/korean1.csv')
df_ko2 = pd.read_csv('embeddings/korean2.csv')

df_hi1 = pd.read_csv('embeddings/hindi1.csv')
df_hi2 = pd.read_csv('embeddings/hindi2.csv')

df_sp1 = pd.read_csv('embeddings/spanish1.csv')
df_sp2 = pd.read_csv('embeddings/spanish2.csv')

df_vi1 = pd.read_csv('embeddings/vietnamese1.csv')
df_vi2 = pd.read_csv('embeddings/vietnamese2.csv')


df_ar1 = pd.read_csv('embeddings/arabic1.csv')
df_ar2 = pd.read_csv('embeddings/arabic2.csv')
```

```python
# concatening them one after the other along the vertical axis ; axis = 0
df_all = pd.concat([df_ch1, df_ch2,df_ko1,df_ko2,df_hi1,df_hi2,df_sp1,df_sp2,df_vi1,df_vi2,df_ar1,df_ar2], axis=0)
```

```python
# idk some xtra column is coming so deleting it

df_all = df_all.drop(columns=['Unnamed: 0'])
df_all
```

```python
df_all.drop(columns=['Y'])
```

```python
df_all['Y']
```

```python
# now splitting the df according to features and labels
X = df_all.drop(columns=['Y'])
y = df_all['Y']
X
```

```python
y
```

```python
#labels are in string form so encoding them to get numerical labels
le = LabelEncoder()
y_encoded = le.fit_transform(y)
```

```python
# spiltting the data into train and test with 80-20 respectively

X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y_encoded, 
    test_size=0.2, 
    random_state=42, 
    stratify=y_encoded
)
```

```python
#setting the parameters
params = {
    'objective': 'multi:softprob',
    'num_class': len(le.classes_), 
    'max_depth': 3,
    'learning_rate': 0.05,         
    'n_estimators': 2000,          
    'subsample': 0.8,
    'colsample_bytree': 0.8,
    'min_child_weight': 30,        
    'gamma': 0.2,                  
    'alpha': 10,
    'reg_lambda': 5,               
    'random_state': 42,
    'eval_metric': 'mlogloss',
    'early_stopping_rounds': 20    
}

#initializing the model
model = XGBClassifier(**params)

# training the model
model.fit(
    X_train, y_train,
    eval_set=[(X_test, y_test)], 
    verbose=10  # Print progress every 10 rounds so it's less noisy
)

# getting the accuracy
train_acc = model.score(X_train, y_train)
test_acc = model.score(X_test, y_test)

y_pred_numeric = model.predict(X_test)

y_pred_languages = le.inverse_transform(y_pred_numeric)
y_test_languages = le.inverse_transform(y_test)

print(f"New Train Accuracy: {train_acc}")
print(f"New Test Accuracy: {test_acc}")
```

```python
print(f"New Train Accuracy: {train_acc}")
print(f"New Test Accuracy: {test_acc}")
```

```python
y_pred_languages
```

```python
print("\nClassification Report:")
print(classification_report(y_test_languages, y_pred_languages))
```

```python
plt.figure(figsize=(20, 20))
cm = confusion_matrix(y_test_languages, y_pred_languages)
sns.heatmap(cm, annot=False, xticklabels=le.classes_, yticklabels=le.classes_)
plt.show()
```
