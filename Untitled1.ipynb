```python
import os
import numpy as np
from tqdm import tqdm
import pandas as pd
```

```python
NEW_EMBED_DIR = 'new_embeddings/Vietnamese/wav_THV/'
OUTPUT_DIR = 'new_embeddings/'
```

```python
os.system(f"ls {NEW_EMBED_DIR} | grep '\.npy$' | wc -l")
```

```python
data = []
labels = []
count = 0
with os.scandir(NEW_EMBED_DIR) as d:
    for e in tqdm(d):
        #print(e.name)
        if e.is_file() and e.name.endswith('.npy'):
            count += 1
            embd = np.load(f"{NEW_EMBED_DIR}/{e.name}")
            labels.append('Vietnamese')
            data.append(embd)
print(count)
```

```python
df = pd.DataFrame(data)
df_labels = pd.DataFrame(labels)
```

```python
df_labels.columns = ['Y']
df_labels
```

```python
df_full = pd.concat([df, df_labels], axis=1)
df_full
```
